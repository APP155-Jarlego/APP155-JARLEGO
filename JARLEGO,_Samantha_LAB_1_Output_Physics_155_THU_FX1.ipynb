Figure 1: f(x) = x^2

```python
import numpy as np
import matplotlib.pyplot as plt
i = np.arange(1, 101)
x = 0.1 * i

plt.figure(figsize=(8, 4))
plt.plot(x, x**2, label=r'$f(x) = x^2$', color='orange')
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(r"Figure 1: $f(x) = x^2$")
plt.grid(True)
plt.show
```

Figure 2: f(x) = cos(x) and f(x) = 1-cos(x)

```python
import numpy as np
import matplotlib.pyplot as plt
i = np.arange(1, 101)
x = 0.1 * i

plt.figure(figsize=(8, 4))
plt.plot(x, np.cos(x), label='f(x) = cos(x)', color='green')
plt.plot(x, 1-np.cos(x), label='f(x) = 1-cos(x)', color='purple')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Figure 2: f(x) = cos(x) and f(x) = 1-cos(x)')
plt.legend()
plt.grid(True)
plt.show
```

Figure 3: f(x) = (1-cos(x)) / x^2

```python
import numpy as np
import matplotlib.pyplot as plt
i = np.arange(1, 101)
x = 0.1 * i
y_c = (1 - np.cos(x)) / (x**2)

plt.figure(figsize=(8, 4))
plt.plot(x, y_c, label=r'$f(x) = \frac{1-cos(x)}{x^2}$', color='magenta')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title(r'Figure 3: $f(x) = \frac{1-\cos(x)}{x^2}$')
plt.legend()
plt.grid(True)
plt.show
```
