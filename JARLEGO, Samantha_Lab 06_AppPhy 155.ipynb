```python
#Part 1
import numpy as np

def testcreate(n, val):
    A = np.arange(val, val+n*n).reshape(n, n)
    A = np.sqrt(A)
    return A

def mag(xs):
    return np.sqrt(np.sum(xs*xs))

def power(A, z0, eps=1.e-6, kmax=1000, verbose=False):
    qold = z0/mag(z0) #q^(0)

    converged = False
    for k in range(1, kmax):
        zs = A@qold #z^(k) = A q^(k-1)
        qnew = zs/mag(zs) #q^(k) = z^(k)/||z^(k)||

#stopping criterion
        err = np.sum(np.abs((qnew - qold)/qnew))
        if err <= eps:
            converged = True
            break

        qold = qnew

    if not converged:
        print("Warning: Max Number of Iterations Reached")

    lam = qnew@A@qnew
    return lam, qnew, k

#testing on matrix A using n=4, val= 21
A = testcreate(4, 21)
print(A)

z0 = np.ones(4)

for eps in [1.e-2, 1.e-4, 1.e-6, 1.e-10]:
    lam, qs, k = power(A, z0, eps)
    print("eps =", eps, " iterations =", k)
    print(lam)
    print(qs)

npeigvals, npeigvecs = np.linalg.eig(A)
print(npeigvals[0])
print(npeigvecs[:, 0])
```

```python
#Part 2
import numpy as np
M = 12
a = 1
I = (M**a**2/12)*np.array([[8, -3, -3],
                           [-3, 8, -3],
                           [-3, -3, 8]])

#2a.)
z0 = np.array([1.0, 1.0, 1.0])
lam, qs, k = power(I, z0, 1.e-10)
print(lam)
print(qs)
print(k)

#2b.)
npeigvals, npeigvecs = np.linalg.eig(I)
print(npeigvals)
print(npeigvecs)
```
