```python
import numpy as np
import matplotlib.pyplot as plt

def f_original (x):
  return (1.0 - np.cos(x)) / x**2

def f_rewritten (x):
  return 2.0 * np.sin(x/2.0)**2 / x**2

def f_first_exact (x):
  return (x * np.sin(x) - 2.0 * (1.0 - np.cos(x))) / x**3

def f_second_exact (x):
  return ((x**2 - 6.0) * np.cos(x) - 4.0 * x * np.sin(x) + 6.0) / x**4

def cd_first(func, x, h):
  return (func(x+h) - func(x-h)) / (2.0 * h)

def cd_second(func, x, h):
  return (func(x+h) - 2.0 * func(x) + func(x-h)) / (h**2)



#Part (a)
x_a = 0.004
f_first_at_a = f_first_exact (x_a)
print("(a) f'(0.004) [exact]=", f_first_at_a)

#Part (b)
hs_1 = np.array([10.0**(-n) for n in range (1, 7)])
err_direct_1 = np.array(
    [abs(cd_first(f_original, x_a, h) - f_first_at_a) for h in hs_1]
  )

plt.figure()
plt.loglog(hs_1, err_direct_1, 'o-', label='f(x): naive (1-cos x)/x^2')
plt.xlabel('h')
plt.ylabel('Absolute Error')
plt.title('(b) Error in the Central-Difference 1st Derivative at x=0.004')
plt.grid(True, which= 'both')
plt.legend()
plt.show()

#Part (c)
err_stable_1 = np.array(
    [abs(cd_first(f_rewritten, x_a, h) - f_first_at_a) for h in hs_1]
)

plt.figure()
plt.loglog(hs_1, err_direct_1, 'o-', label='f(x): naive form')
plt.loglog(hs_1, err_stable_1, 's-', label='g(x): rewritten stable form')
plt.xlabel('h')
plt.ylabel('Absolute Error')
plt.title('(c) Naive vs Stable Form')
plt.grid(True, which= 'both')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def f_original (x):
  return (1.0 - np.cos(x)) / x**2

def f_rewritten (x):
  return 2.0 * np.sin(x/2.0)**2 / x**2

def f_first_exact (x):
  return (x * np.sin(x) - 2.0 * (1.0 - np.cos(x))) / x**3

def f_second_exact (x):
  return ((x**2 - 6.0) * np.cos(x) - 4.0 * x * np.sin(x) + 6.0) / x**4

def cd_first(func, x, h):
  return (func(x+h) - func(x-h)) / (2.0 * h)

def cd_second(func, x, h):
  return (func(x+h) - 2.0 * func(x) + func(x-h)) / (h**2)


#Part (d)
x_d = 2.0
f_second_at_d = f_second_exact (x_d)
print("(d) f''(2.0) [exact]=", f_second_at_d)

#Part (e)
hs_2 = np.array([10.0**(-n) for n in range (1, 8)])
err_direct_2 = np.array(
    [abs(cd_second(f_original, x_d, h) - f_second_at_d) for h in hs_2]
  )

plt.figure()
plt.loglog(hs_2, err_direct_2, 'o-', label='f(x): naive (1-cos x)/x^2')
plt.xlabel('h')
plt.ylabel('Absolute Error')
plt.title('(e) Error in the Central-Difference 2nd Derivative at x=2.0')
plt.grid(True, which= 'both')
plt.legend()
plt.show()

#Part (f)
err_stable_2 = np.array(
    [abs(cd_second(f_rewritten, x_d, h) - f_second_at_d) for h in hs_2]
)

plt.figure()
plt.loglog(hs_2, err_direct_2, 'o-', label='f(x): naive form')
plt.loglog(hs_2, err_stable_2, 's-', label='g(x): rewritten stable form')
plt.xlabel('h')
plt.ylabel('Absolute Error')
plt.title('(f) Naive vs Stable Form')
plt.grid(True, which= 'both')
plt.legend()
plt.show()
```
