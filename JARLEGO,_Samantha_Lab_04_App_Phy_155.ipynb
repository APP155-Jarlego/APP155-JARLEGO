**1. Idiomatic NumPy**

```python
import numpy as np

def sumofints(nmax):
  return np.sum(np.arange(1, nmax+1))

for nmax in (10, 100):
  x = sumofints(nmax)
  print(nmax, x)
```

**2. Matrices**

Code Block [2]



```python
import numpy as np

def testcreate(n, val):
  A = np.arange(val, val+n*n).reshape(n, n)
  A = np.sqrt(A)
  return A

for n,val in [(4, 21), (10,21), (20,21)]:
  A = testcreate(n, val)
  manual_frobenius = np.sqrt(np.sum(A**2))
  builtin_frobenius = np.linalg.norm(A)
  print(n, val, manual_frobenius, builtin_frobenius)
```

Notes on Code Block [2]
* By definition, *sqrt(sum of
squares of all entries)* is the same as those computed manually.

* *A**2* undoes the np.sqrt(A), so the frobenius norm reduces to the *sqrt(sum of
the original arange valuess)*

Code Block [3]

```python
import numpy as np

for n, val in [(4, 21), (10, 21), (20, 21)]:
  A = testcreate(n, val)
  print(n, val, np.linalg.det(A))
```

Notes on Code Block [3]
* The two quantities behave differently.

* As depictede in the output, the Frobenius norm grows smoothly and moderately with 'n'.

* The determinant collapses toward 0 at an accelerating rate. This is because of the *testcreate* command, which builds each row by taking consecutive integers and square-rooting them (as 'n' grows, rows become more linearly dependent)

* Norm: tells us the matrix size.

* Determinant: tells us the matrix is becoming nearly singular as n increases.
